```python
%%writefile product_validator.l
%{
#include <stdio.h>
%}

%%
^[A-Z]{2,}-?[0-9]{4}[a-zA-Z]*\r?\n    { printf("Valid\n"); }
^[A-Z]{2,}-?[0-9]{4}[a-zA-Z]*$        { printf("Valid\n"); }
^[^\r\n]*\r?\n                        { printf("Invalid\n"); }
^[^\r\n]+$                            { printf("Invalid\n"); }
%%

int yywrap(void) {
    return 1;
}

int main(void) {
    yylex();
    return 0;
}
```

```python
!lex product_validator.l
!gcc lex.yy.c -o product_validator -lfl
!./product_validator
```
