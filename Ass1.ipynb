```python
%%writefile lexer.l
%{
#include <stdio.h>
%}

%%
^[A-Z]+[0-9]?[a-z]*\r?\n    { printf("Valid\n"); }
^[A-Z]+[0-9]?[a-z]*$        { printf("Valid\n"); }
^[^\r\n]*\r?\n              { printf("Invalid\n"); }
^[^\r\n]+$                  { printf("Invalid\n"); }
%%

int yywrap(void) {
    return 1;
}

int main(void) {
    yylex();
    return 0;
}
```

```python
!lex log_validator.l
!gcc lex.yy.c -o log_validator -lfl
!./log_validator
```
